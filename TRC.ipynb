```python
import random

def teorema_del_residuo_chino(remainders, moduli):
    if len(remainders) != len(moduli):
        raise ValueError("El tamaño de remainders tiene que ser igual al de moduli.")
    
    total_modulus = 1
    for m in moduli:
        total_modulus *= m
        
    x = 0
    for a_i, m_i in zip(remainders, moduli):
        N_i = total_modulus // m_i
        y_i = pow(N_i, -1, m_i)
        x += a_i * N_i * y_i
        
    return x % total_modulus, total_modulus
```
